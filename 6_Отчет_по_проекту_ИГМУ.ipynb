### 2. Ключевые аналитические выводы по данным (2016–2024)

#### 2.1. Эпидемиологическая ситуация в Иркутской области
- **Первичная заболеваемость снизилась на 49,3%** (с 108,5 до 55,0 на 100 тыс.).
- **Общая заболеваемость (распространённость) снизилась на 63,8%** (с 262,1 до 95,0 на 100 тыс.).
- **Смертность от туберкулёза снизилась** (с 14,2 до 5,9 на 100 тыс., точные значения из дашборда).
- **Ни одного года роста** – устойчивый нисходящий тренд.
- **Сокращение разрыва между общей и первичной заболеваемостью** с 2,4 до 1,7 раза – уменьшение резервуара инфекции.

#### 2.2. Структура и качественные индикаторы
- **Доля внелёгочного туберкулёза** стабильно низкая (около 2–3%).
- **Доля первичной МЛУ** не превышает 8%, но требует мониторинга (цель – снижение).
- **Клиническое излечение** выросло до 82%, **абацилирование** – до 76%.
- **Охват флюорографией взрослых** достиг 72%, но выявляемость методом флюорографии снижается – ранний сигнал возможного роста заболеваемости в будущем.

#### 2.3. Ресурсы и доступность
- **Обеспеченность фтизиатрами** – 0,8 на 10 тыс. (в норме).
- **Укомплектованность физическими лицами** – 68% (дефицит кадров).
- **Обеспеченность койками для взрослых** – 2,1 на 10 тыс. (достаточно).

#### 2.4. Ключевая корреляция (выявлена в EDA)
- **Охват госпитализацией впервые выявленных** сильно коррелирует со снижением заболеваемости (r = -0,92), смертности (r = -0,97) и ростом излечения (r = +0,85). Это **главный управляемый рычаг**.

#### 2.5. Аномалии
- В 2016 году зафиксированы выбросы заболеваемости в возрастных группах 7–14 и 15–17 лет (z-score >2). Возможная причина – активное выявление или изменение методов диагностики.

### 3. Управленческие рекомендации (для главного врача и зав. оргметодотделом)

1. **Поддерживать высокий охват госпитализацией впервые выявленных** (цель – 100%). Это наиболее эффективная мера для снижения заболеваемости и смертности.
2. **Усилить контроль за профосмотрами** – падающая выявляемость флюорографией (с 3,4 до 0,5 на 1000 осмотренных) требует анализа причин (качество оборудования, охват населения).
3. **Сфокусироваться на лечении МЛУ** – доля вторичной МЛУ растёт, что указывает на ошибки в схемах терапии. Провести аудит лечения пациентов с устойчивостью.
4. **Обратить внимание на детскую заболеваемость** – хотя абсолютные значения низкие, необходимо усилить туберкулинодиагностику и вакцинацию.
5. **Использовать конструкторы дашборда** для самостоятельного анализа заведующим оргметодотделом – гибкая сборка показателей позволит углублённо изучать проблемные зоны.
