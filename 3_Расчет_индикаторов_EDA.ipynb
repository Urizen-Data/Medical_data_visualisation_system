# Часть 3. Расчет индикаторов ИГМУ


## Аннотация к расчёту показателей

Представленный перечень включает **показатели** для эпидемиологического мониторинга туберкулёза на основе утверждённых форм федерального статистического наблюдения (№8, №33, №30). Расчёты производятся с использованием **среднегодовой численности населения** (общей, по полу и возрастным группам) Иркутской области.  
Все интенсивные показатели приведены на **100 тыс. населения** (если не указано иное), относительные – в процентах, выявляемость – на 1000 осмотренных, обеспеченность – на 10 000 населения.  
Показатели сгруппированы в **9 тематических блоков**, соответствующих разделам отчётных форм и этапам оказания противотуберкулёзной помощи (выявление, учёт, лечение, профилактика, кадры, койки). Данные предназначены для построения дашбордов, факторного анализа и управленческих решений.

---

## 9 блоков показателей

**Блок 1. 📊 Форма №8, табл.1000 – Первичная заболеваемость по полу и возрастам**  
Расчёт числа **впервые выявленных** больных активным туберкулёзом (на 100 тыс.) с детализацией по полу и 10 возрастным группам (от 0–4 до 65+ лет), включая отдельные показатели для мужчин и женщин в каждой возрастной группе.

**Блок 2. 🧬 Форма №33, табл.2100 – Основные эпидпоказатели и ТБ+ВИЧ**  
Охватывает **первичную (ПЗ)** и **общую заболеваемость (ОЗ)** туберкулёзом (все формы, органов дыхания, лёгких, ФКТ, с распадом, внелёгочный), инвалидность, охват тестированием на ВИЧ, а также показатели **ТБ+ВИЧ** (заболеваемость, смертность).

**Блок 3. 🔍 Форма №33, табл.2200, 2310 – Выявление, смертность, посмертная диагностика**  
Показатели **выявления** (по данным медосмотров и туберкулинодиагностики, доля методов), **посмертная диагностика**, **смертность** от туберкулёза и от других причин, доля умерших на учёте менее года.

**Блок 4. 💊 Форма №33, табл.2500 – Лекарственная устойчивость (МЛУ) и бактериовыделение**  
Оценивает **бактериовыделение** (ПЗ туберкулёза ОД с МБТ+, доля среди впервые выявленных), охват исследованиями на **МЛУ**, частоту и распространённость МЛУ, долю **первичной и вторичной МЛУ** МБТ.

**Блок 5. 🩺 Форма №33, табл.2300, 2400 – Эффективность лечения**  
Включает **частоту рецидивов** (всего и из III группы), **охват химиопрофилактикой** взрослых из контакта с бактериовыделителями, **клиническое излечение** больных туберкулёзом ОД и **абацилирование**.

**Блок 6. 🏥 Форма №33, табл.2600 – Госпитализация**  
Показатели **охвата госпитализацией** (впервые выявленных всех, с МБТ+, контингентов с МБТ+) и **госпитализированной заболеваемости** (всего, впервые выявленных, детей, подростков).

**Блок 7. 📋 Форма №30, табл.2513 – Охват осмотрами и выявляемость**  
Характеризует **профилактическую работу**: охват населения всеми видами осмотров, охват иммунодиагностикой детей 0–14 лет, охват флюорографией населения старше 15 лет, выявляемость туберкулёза методом флюорографии и бактериоскопии (на 1000 осмотренных).

**Блок 8. 👨‍⚕️ Форма №30, табл.1100 – Кадры (фтизиатры)**  
Показатели обеспеченности, укомплектованности и коэффициента совместительства фтизиатрами (всего и участковыми).

**Блок 9. 🛏️ Форма №30, табл.3100 – Койки**  
Показатели обеспеченности туберкулёзными койками для взрослых и детей, а также госпитализированной заболеваемости взрослого и детского населения.

```python
# ----------------------------------------------------------------------
# 1. НАСТРОЙКА ПУТИ К ДАННЫМ
# ----------------------------------------------------------------------
data_dir = r"C:\Users\urize\Ирткутск_туберкулез"

# ----------------------------------------------------------------------
# 2. ФУНКЦИЯ БЕЗОПАСНОЙ ЗАГРУЗКИ ФАЙЛОВ
# ----------------------------------------------------------------------
def safe_load(filename):
    try:
        return pd.read_csv(os.path.join(data_dir, filename))
    except FileNotFoundError:
        print(f"Предупреждение: файл {filename} не найден, создан пустой DataFrame")
        return pd.DataFrame()

# ----------------------------------------------------------------------
# 3. ЗАГРУЗКА ВСЕХ НЕОБХОДИМЫХ ДАТАФРЕЙМОВ
# ----------------------------------------------------------------------
df_all_1000 = safe_load("df_all_1000.csv")
df_all_2800 = safe_load("df_all_2800.csv")
df_all_2700 = safe_load("df_all_2700.csv")
df_all_2100 = safe_load("df_all_2100.csv")
df_all_2200 = safe_load("df_all_2200.csv")
df_all_2300 = safe_load("df_all_2300.csv")
df_all_2400 = safe_load("df_all_2400.csv")
df_all_2500 = safe_load("df_all_2500.csv")
df_all_2600 = safe_load("df_all_2600.csv")
df_all_2310 = safe_load("df_all_2310.csv")
df_all_2513 = safe_load("df_all_2513.csv")
df_all_1100 = safe_load("df_all_1100.csv")
df_all_3100 = safe_load("df_all_3100.csv")

irkutsk_population_avg = safe_load("irkutsk_population_avg.csv")

# ----------------------------------------------------------------------
# 4. СОЗДАНИЕ СЛОВАРЯ ДЛЯ ДОСТУПА К ДАТАФРЕЙМАМ ПО НОМЕРУ ТАБЛИЦЫ
# ----------------------------------------------------------------------
df_dict = {}
for name, df in zip(['1000','2100','2200','2300','2400','2500','2600','2700','2800','2310','2513','1100','3100'],
                    [df_all_1000, df_all_2100, df_all_2200, df_all_2300, df_all_2400, df_all_2500, df_all_2600,
                     df_all_2700, df_all_2800, df_all_2310, df_all_2513, df_all_1100, df_all_3100]):
    if not df.empty:
        df_dict[name] = df

# ----------------------------------------------------------------------
# 5. ПОДГОТОВКА ТАБЛИЦЫ НАСЕЛЕНИЯ
# ----------------------------------------------------------------------
pop = irkutsk_population_avg.copy()

# ----------------------------------------------------------------------
# 6. ФУНКЦИЯ ПОЛУЧЕНИЯ ЗНАЧЕНИЯ ИЗ ТАБЛИЦЫ ПО КООРДИНАТАМ
# ----------------------------------------------------------------------
def get_value(table: str, year: int, row_spec, col_spec):
    df = df_dict.get(table)
    if df is None or df.empty:
        return None
    
    try:
        if isinstance(row_spec, str) and '+' in row_spec:
            rows = [int(x) for x in row_spec.split('+')]
        else:
            rows = [int(row_spec)]
        
        if isinstance(col_spec, str) and '+' in col_spec:
            cols = [int(x) for x in col_spec.split('+')]
        else:
            cols = [int(col_spec)]
        
        total = 0
        found = False
        
        for r in rows:
            for c in cols:
                mask = (df['Год'] == year) & (df['Строка'] == r) & (df['Графа'] == c)
                val = df.loc[mask, 'Значение']
                if not val.empty:
                    total += val.iloc[0]
                    found = True
        
        return total if found else None
    
    except Exception as e:
        print(f"Ошибка в get_value({table},{year},{row_spec},{col_spec}): {e}")
        return None

# ----------------------------------------------------------------------
# 7. ФУНКЦИЯ ПОЛУЧЕНИЯ ЧИСЛЕННОСТИ НАСЕЛЕНИЯ
# ----------------------------------------------------------------------
def get_population(year: int, denom: str):
    if pop.empty:
        return 1
    
    # Прямое соответствие: ищем колонку с таким же именем
    # Если колонка есть в pop, берём её
    if denom in pop.columns:
        try:
            val = pop.loc[pop['Год'] == year, denom]
            return val.iloc[0] if not val.empty else 1
        except:
            return 1
    
    # Если точного совпадения нет, используем маппинг
    mapping = {
        'Всего': 'Всего',
        'Мужчины': 'м_Всего',
        'Женщины': 'ж_Всего',
        '0-4': '0-4_лет',
        'м_0-4': 'м_0-4_лет',
        'ж_0-4': 'ж_0-4_лет',
        '5-6': '5-6_лет',
        'м_5-6': 'м_5-6_лет',
        'ж_5-6': 'ж_5-6_лет',
        '7-14': '7-14_лет',
        'м_7-14': 'м_7-14_лет',
        'ж_7-14': 'ж_7-14_лет',
        '0-14': '0-14_лет',
        'м_0-14': 'м_0-14_лет',
        'ж_0-14': 'ж_0-14_лет',
        '15-17': '15-17_лет',
        'м_15-17': 'м_15-17_лет',
        'ж_15-17': 'ж_15-17_лет',
        '0-17': '0-17_лет',
        'м_0-17': 'м_0-17_лет',
        'ж_0-17': 'ж_0-17_лет',
        '15+': '15+',
        'м_15+': 'м_15+',
        'ж_15+': 'ж_15+',
        '18+': '18+',
        'м_18+': 'м_18+',
        'ж_18+': 'ж_18+',
        '18-24': '18-24_лет',
        'м_18-24': 'м_18-24_лет',
        'ж_18-24': 'ж_18-24_лет',
        '25-34': '25-34_лет',
        'м_25-34': 'м_25-34_лет',
        'ж_25-34': 'ж_25-34_лет',
        '35-44': '35-44_лет',
        'м_35-44': 'м_35-44_лет',
        'ж_35-44': 'ж_35-44_лет',
        '45-54': '45-54_лет',
        'м_45-54': 'м_45-54_лет',
        'ж_45-54': 'ж_45-54_лет',
        '55-64': '55-64_лет',
        'м_55-64': 'м_55-64_лет',
        'ж_55-64': 'ж_55-64_лет',
        '65+': '65+',
        'м_65+': 'м_65+',
        'ж_65+': 'ж_65+',
    }
    
    col = mapping.get(denom, denom)
    
    try:
        val = pop.loc[pop['Год'] == year, col]
        return val.iloc[0] if not val.empty else 1
    except:
        return 1

# ----------------------------------------------------------------------
# 8. ФУНКЦИЯ ПОЛУЧЕНИЯ ТЕКСТОВОГО НАЗВАНИЯ ПОКАЗАТЕЛЯ
# ----------------------------------------------------------------------
def get_indicator_name(table: str, row_spec, col_spec):
    df = df_dict.get(table)
    if df is None or df.empty:
        return f"[Таблица {table} отсутствует]"
    
    try:
        if isinstance(row_spec, str) and '+' in row_spec:
            row = int(row_spec.split('+')[0])
        else:
            row = int(row_spec)
        
        exclude = ['Год', 'Значение', 'Строка', 'Графа', 'Таблица', 
                   'Код по МКБ-Х пересмотра', 'Код по МКБ-X пересмотра']
        
        text_cols = [c for c in df.columns if c not in exclude and df[c].dtype == 'object']
        
        mask = (df['Строка'] == row) & (df['Графа'] == col_spec)
        match = df.loc[mask]
        
        if not match.empty:
            parts = [str(match.iloc[0][c]).strip() for c in text_cols if pd.notna(match.iloc[0][c])]
            return " | ".join(parts) if parts else f"Строка {row_spec}, графа {col_spec}"
        
        return f"Строка {row_spec}, графа {col_spec}"
    
    except Exception as e:
        return f"Ошибка: {e}"

# ----------------------------------------------------------------------
# 9. СПИСОК ГОДОВ ДЛЯ АНАЛИЗА
# ----------------------------------------------------------------------
years = list(range(2016, 2025))

# ----------------------------------------------------------------------
# 10. ПРОВЕРКА ЗАГРУЗКИ
# ----------------------------------------------------------------------
print("Загрузка и функции готовы")
print(f"Таблица населения загружена, размер: {pop.shape}")
print(f"Колонки таблицы населения: {pop.columns.tolist()}")
```

### Проверка indicator = numerator / denominator

```python
# Проверяем, что indicator = numerator / denominator * multiplier
# Множитель извлекаем из формулы

print("=== ПРОВЕРКА РАСЧЁТА indicator ===\n")

def get_multiplier_from_formula(formula):
    """Извлекает множитель из строки формулы"""
    if pd.isna(formula):
        return 1
    # Ищем число после '*' в конце формулы
    import re
    match = re.search(r'\*\s*(\d+(?:\.\d+)?)$', str(formula))
    if match:
        return float(match.group(1))
    return 1

df_check = df_indicators.copy()

# Извлекаем множитель
df_check['multiplier'] = df_check['formula'].apply(get_multiplier_from_formula)

# Пересчитываем
df_check['indicator_calc'] = df_check['numerator'] / df_check['denominator'] * df_check['multiplier']
df_check['indicator_calc'] = df_check['indicator_calc'].round(2)

# Сравниваем
df_check['diff'] = (df_check['indicator'] - df_check['indicator_calc']).round(2)
df_check['match'] = df_check['indicator'] == df_check['indicator_calc']

print(f"Совпадает: {df_check['match'].sum()} из {len(df_check)}")

mismatch = df_check[~df_check['match']]
if len(mismatch) > 0:
    print("\nНЕСОВПАДЕНИЯ:")
    print(mismatch[['id', 'short_name', 'year', 'indicator', 'numerator', 'denominator', 'multiplier', 'indicator_calc', 'diff']].to_string(index=False))
else:
    print("\n✅ Все значения совпадают!")
```

### Блок i. Расчет индекса хронизации

```python
import pandas as pd
import numpy as np

# Приводим indicator к числу
df_indicators_sub['indicator'] = pd.to_numeric(df_indicators_sub['indicator'], errors='coerce')

# Работаем с копией для замены пустых значений
df_work = df_indicators_sub.copy()
df_work['disease'] = df_work['disease'].replace('', 'ТБ всего')
df_work['population'] = df_work['population'].replace('', 'общ.')

# 1. Выделяем первичную и общую заболеваемость
primary = df_work[
    (df_work['type'] == 'Заболеваемость') & 
    (df_work['subtype'] == 'первичная')
].copy()

general = df_work[
    (df_work['type'] == 'Заболеваемость') & 
    (df_work['subtype'] == 'общая')
].copy()

# 2. Объединяем по year, disease, population
merged = pd.merge(
    primary[['year', 'disease', 'population', 'indicator', 'name', 'short_name']],
    general[['year', 'disease', 'population', 'indicator', 'name', 'short_name', 'id']],
    on=['year', 'disease', 'population'],
    suffixes=('_prim', '_gen')
)

# 3. Очистка и расчёт индекса
merged = merged.dropna(subset=['indicator_prim', 'indicator_gen'])
merged = merged[merged['indicator_prim'] > 0]
merged['chron_index'] = merged['indicator_gen'] / merged['indicator_prim']

if merged.empty:
    print("Нет данных для расчёта хронизации (нет пар первичная/общая).")
else:
    # 4. Генерация новых id на основе id общей заболеваемости
    existing_ids = set(df_indicators_sub['id'].tolist())
    new_records = []
    
    for _, row in merged.iterrows():
        # Берём id общей заболеваемости (например, id_2.1.4)
        gen_id = row['id']
        # Заменяем 'id_2' на 'id_10'
        if gen_id.startswith('id_2'):
            base_new_id = 'id_10' + gen_id[4:]   # убираем 'id_2' -> оставляем '.1.4' и добавляем 'id_10'
        else:
            # fallback (на случай, если id не начинается с id_2)
            base_new_id = 'id_10.' + gen_id.split('.', 1)[1] if '.' in gen_id else 'id_10.' + gen_id
        
        new_id = base_new_id
        # Проверяем уникальность, если уже существует – добавляем суффикс .1, .2...
        suffix = 1
        while new_id in existing_ids:
            new_id = f"{base_new_id}.{suffix}"
            suffix += 1
        existing_ids.add(new_id)
        
        # 5. Формируем name на основе готового наименования общей заболеваемости
        orig_name = row['name_gen']
        if 'Показатель общей заболеваемости' in orig_name:
            new_name = orig_name.replace('Показатель общей заболеваемости', 'Хронизация заболеваемости')
        elif 'Показатель первичной заболеваемости' in orig_name:
            new_name = orig_name.replace('Показатель первичной заболеваемости', 'Хронизация заболеваемости')
        else:
            new_name = f"Хронизация заболеваемости {orig_name}"
        
        # Аналогично для short_name
        orig_short = row['short_name_gen']
        if 'общей заболеваемости' in orig_short:
            new_short = orig_short.replace('общей заболеваемости', 'хронизации')
        elif 'первичной заболеваемости' in orig_short:
            new_short = orig_short.replace('первичной заболеваемости', 'хронизации')
        else:
            new_short = f"Хронизация {orig_short}"
        new_short = new_short[:50]
        
        new_rec = {
            'id': new_id,
            'name': new_name,
            'short_name': new_short,
            'year': row['year'],
            'indicator': row['chron_index'],
            'numerator': None,
            'denominator': None,
            'formula': 'общая заболеваемость / первичная заболеваемость',
            'source': 'расчетный (на основе данных)',
            'tables': None,
            'type': 'Хронизация',
            'subtype': 'общ.',
            'disease': row['disease'],
            'population': row['population']
        }
        new_records.append(new_rec)
    
    # 6. Добавляем в исходный DataFrame
    df_new = pd.DataFrame(new_records)
    df_indicators_sub = pd.concat([df_indicators_sub, df_new], ignore_index=True)
    print(f"✅ Добавлено {len(df_new)} записей с индексом хронизации.")
    
    # Показываем пример
    print("\n📋 Пример добавленных записей (id и name):")
    print(df_new[['id', 'name', 'year', 'indicator', 'disease', 'population']].head())
```

#### Численность населения добавим. pop_per_age

```python
# ============================================================
# 1. ЗАГРУЗКА ДАННЫХ
# ============================================================
# tb_indicators_measure = pd.read_csv('tb_indicators_measure.csv')
irkutsk_population = pd.read_csv('irkutsk_population_2016_2024.csv')

# ============================================================
# 2. ПРЕОБРАЗОВАНИЕ НАСЕЛЕНИЯ В ДЛИННЫЙ ФОРМАТ
# ============================================================
pop_records = []
for _, row in irkutsk_population.iterrows():
    year = row['Год']
    gender = row['Пол']
    for col in irkutsk_population.columns:
        if col in ('Год', 'Пол'):
            continue
        value = row[col]
        if pd.isna(value):
            continue
        pop_records.append({
            'year': year,
            'gender': gender,
            'age_group': col,
            'pop_per_age': value
        })

pop_df = pd.DataFrame(pop_records)

# Создаём отдельный словарь для быстрого доступа к численности всего населения
total_pop_dict = {}
for _, row in pop_df[pop_df['age_group'] == 'Всего'].iterrows():
    total_pop_dict[(row['year'], row['gender'])] = row['pop_per_age']

# ============================================================
# 3. ДОБАВЛЕНИЕ pop_per_age В tb_indicators_measure
# ============================================================
def get_pop_per_age(row):
    year = row['year']
    population = row['population']
    
    # Случай 1: 'всего', 'женщины', 'мужчины' — берём общую численность
    if population in ('всего', 'женщины', 'мужчины'):
        gender = population
        return total_pop_dict.get((year, gender), None)
    
    # Случай 2: '0-4', '5-6', ..., '0-14', '0-17', '15+', '18+'
    # Ищем в pop_df точное совпадение
    gender = 'всего'  # по умолчанию
    age = population
    
    if population.startswith('женщины '):
        gender = 'женщины'
        age = population[len('женщины '):]
    elif population.startswith('мужчины '):
        gender = 'мужчины'
        age = population[len('мужчины '):]
    
    # Ищем в pop_df
    match = pop_df[(pop_df['year'] == year) & 
                   (pop_df['gender'] == gender) & 
                   (pop_df['age_group'] == age)]
    if not match.empty:
        return match.iloc[0]['pop_per_age']
    
    return None

tb_indicators_measure['pop_per_age'] = tb_indicators_measure.apply(get_pop_per_age, axis=1)

# ============================================================
# 4. ПРОВЕРКА
# ============================================================
print("\n=== ПРОВЕРКА: population='всего' ===")
print(tb_indicators_measure[tb_indicators_measure['population'] == 'всего'][['year', 'population', 'pop_per_age']].drop_duplicates().head(10))

print("\n=== ПРОВЕРКА: population='0-4' ===")
print(tb_indicators_measure[tb_indicators_measure['population'] == '0-4'][['year', 'population', 'pop_per_age']].drop_duplicates().head(10))

print("\n=== ПРОВЕРКА: population='женщины 25-34' ===")
print(tb_indicators_measure[tb_indicators_measure['population'] == 'женщины 25-34'][['year', 'population', 'pop_per_age']].drop_duplicates().head(10))

print(f"\nДоля пропусков: {tb_indicators_measure['pop_per_age'].isna().mean()*100:.1f}%")
```

```python

# ============================================================
# 1. Определяем список name, имеющих 'отн. единицы' (относительные на 100 тыс.)
# ============================================================
names_with_rel = tb_indicators_measure[
    tb_indicators_measure['measure_type'] == 'отн. единицы'
]['name'].unique()
print(f"Всего name с 'отн. единицы': {len(names_with_rel)}")

# ============================================================
# 2. Исключаем name, содержащие нежелательные подстроки
# ============================================================
exclude_substrings = ['%', 'на 1000', 'на 100', 'на 10000']
filtered_names = [
    n for n in names_with_rel 
    if not any(sub in n for sub in exclude_substrings)
]
print(f"После исключения: {len(filtered_names)}")
print("Примеры:", filtered_names[:5])

# ============================================================
# 3. Отбираем только абсолютные числа для отфильтрованных name
# ============================================================
df_abs = tb_indicators_measure[
    (tb_indicators_measure['measure_type'] == 'абс. число') &
    (tb_indicators_measure['name'].isin(filtered_names)) &
    (tb_indicators_measure['kids_adult'].notna()) &
    (tb_indicators_measure['pop_per_age'].notna()) & (tb_indicators_measure['pop_per_age'] > 0)
].copy()

print(f"Количество строк с абсолютными числами: {len(df_abs)}")

# ============================================================
# 4. Агрегация по year, type, subtype, disease, kids_adult
# ============================================================
aggregated = df_abs.groupby(
    ['year', 'type', 'subtype', 'disease', 'kids_adult'],
    as_index=False
).agg(
    absolute_value=('value', 'sum'),
    population_sum=('pop_per_age', 'sum')
)

# ============================================================
# 5. Пересчёт относительного показателя на 100 000
# ============================================================
aggregated['relative_per_100k'] = (
    aggregated['absolute_value'] * 100000 / aggregated['population_sum']
)

# ============================================================
# 6. Преобразование в формат с двумя строками
#    measure_type: 'абс. число' и 'отн. единицы'
# ============================================================
# Создаём две копии
df_abs_long = aggregated[['year', 'type', 'subtype', 'disease', 'kids_adult', 'absolute_value', 'population_sum']].copy()
df_abs_long['measure_type'] = 'абс. число'
df_abs_long.rename(columns={'absolute_value': 'value', 'population_sum': 'pop_per_age'}, inplace=True)

df_rel_long = aggregated[['year', 'type', 'subtype', 'disease', 'kids_adult', 'relative_per_100k', 'population_sum']].copy()
df_rel_long['measure_type'] = 'отн. единицы'
df_rel_long.rename(columns={'relative_per_100k': 'value', 'population_sum': 'pop_per_age'}, inplace=True)

# Объединяем
df_indicators_kids_adult = pd.concat([df_abs_long, df_rel_long], ignore_index=True)

# Сортируем для удобства
df_indicators_kids_adult.sort_values(['year', 'type', 'subtype', 'disease', 'kids_adult'], inplace=True)

# ============================================================
# 7. Результат
# ============================================================
print(f"\nРазмер итогового датасета: {df_indicators_kids_adult.shape}")
print("\nУникальные категории kids_adult:")
print(df_indicators_kids_adult['kids_adult'].unique())
print("\nПервые 10 строк:")
print(df_indicators_kids_adult.head(10))

# ============================================================
# 8. Сохранение в CSV и создание кликабельной ссылки для скачивания
# ============================================================
filename = 'df_indicators_kids_adult.csv'
df_indicators_kids_adult.to_csv(filename, index=False, encoding='utf-8-sig')
print(f"✅ CSV сохранён: {filename}")
print(f"   Размер: {df_indicators_kids_adult.shape[0]} строк × {df_indicators_kids_adult.shape[1]} колонок")

# Кликабельная ссылка для скачивания
csv_data = df_indicators_kids_adult.to_csv(index=False, encoding='utf-8-sig')
csv_b64 = base64.b64encode(csv_data.encode()).decode()
display(HTML(f'<a href="data:file/csv;base64,{csv_b64}" download="{filename}">📥 Скачать {filename}</a>'))
```

### Распределение indicator по типам показателей


```python
# 2. Распределение indicator по типам (автоматическая сетка) - с выбором population
# Если нет 'всего', используем '18+'

def get_population_filter(pop_candidates):
    """Возвращает первое существующее значение population из списка кандидатов"""
    for pop in pop_candidates:
        if pop in df_indicators_sub['population'].values:
            return pop
    return None

# Для каждого типа определяем подходящий population
population_for_type = {}
for cat in df_indicators_sub['type'].unique():
    available_pops = df_indicators_sub[df_indicators_sub['type'] == cat]['population'].unique()
    if 'всего' in available_pops:
        population_for_type[cat] = 'всего'
    elif '18+' in available_pops:
        population_for_type[cat] = '18+'
    else:
        # Если нет ни 'всего', ни '18+', берём первый попавшийся
        population_for_type[cat] = available_pops[0] if len(available_pops) > 0 else None

unique_types = df_indicators_sub['type'].unique()
n_types = len(unique_types)
n_cols = min(4, n_types)
n_rows = (n_types + n_cols - 1) // n_cols

fig, axes = plt.subplots(n_rows, n_cols, figsize=(5*n_cols, 4*n_rows))
if n_rows * n_cols > 1:
    axes = axes.flatten()
else:
    axes = [axes]

for i, cat in enumerate(unique_types):
    pop_val = population_for_type.get(cat)
    if pop_val is not None:
        data = df_indicators_sub[(df_indicators_sub['type'] == cat) & 
                                  (df_indicators_sub['population'] == pop_val)]['indicator'].dropna()
        axes[i].hist(data, bins=60, edgecolor='black')
        axes[i].set_title(f"{cat[:30] if len(cat) <= 30 else cat[:27] + '...'}\n(population: {pop_val})", fontsize=10)
    else:
        axes[i].text(0.5, 0.5, f'Нет данных для {cat}', ha='center', va='center')
        axes[i].set_title(cat[:30] if len(cat) <= 30 else cat[:27] + '...')
    axes[i].set_xlabel('Значение')
    axes[i].set_ylabel('Частота')

for j in range(i+1, len(axes)):
    axes[j].set_visible(False)

plt.tight_layout()
plt.show()
```

### Проверка согласованности расчётов (indicator vs numerator/denominator)

```python
# 3. Проверка согласованности расчётов
def extract_multiplier(formula):
    import re
    match = re.search(r'\*\s*(\d+(?:\.\d+)?)$', str(formula))
    if match:
        return float(match.group(1))
    return 1

df_check = df_indicators_sub.copy()
df_check['multiplier'] = df_check['formula'].apply(extract_multiplier)
df_check['calculated'] = df_check['numerator'] / df_check['denominator'] * df_check['multiplier']
df_check['diff'] = df_check['indicator'] - df_check['calculated']
mismatch = df_check[df_check['diff'].abs() > 0.01]
print(f"\nКоличество несовпадений indicator с расчётом: {len(mismatch)}")
if len(mismatch) > 0:
    print(mismatch[['id', 'year', 'indicator', 'calculated', 'diff', 'numerator', 'denominator', 'multiplier']].head())
```

### Матрица корреляций

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Список ID показателей для корреляционного анализа
corr_ids = [
    'id_1.1',      # ПЗ всего
    'id_2.1.4',    # ОЗ всего
    'id_3.8',      # Смертность от ТБ
    'id_5.4',      # Клиническое излечение
    'id_5.5',      # Абацилирование
    'id_4.2',      # Частота МЛУ
    'id_6.1',      # Охват госпитализацией ВВ
    'id_7.3',      # Охват ФЛГ (15+)
    'id_2.4.4',    # Доля ФКТ среди ВВ
    'id_2.5.4',    # Доля распада среди ВВ
    'id_2.9.1',    # Охват тестированием на ВИЧ (ВВ)
    'id_2.10.1',   # ПЗ ТБ+ВИЧ
]

# 2. Отбираем только те ID, которые есть в данных
available_corr_ids = [id_ for id_ in corr_ids if id_ in df_indicators_sub['id'].values]

if len(available_corr_ids) >= 2:
    # 3. Берём данные по этим ID, убираем дубликаты (год + id)
    corr_data = df_indicators_sub[df_indicators_sub['id'].isin(available_corr_ids)]\
                                 .drop_duplicates(subset=['year', 'id'])
    
    # 4. Широкая таблица: строки = годы, столбцы = id, значения = indicator
    wide_corr = corr_data.pivot(index='year', columns='id', values='indicator')
    wide_corr = wide_corr.dropna(axis=0, how='any')  # удаляем годы с пропусками
    
    if wide_corr.shape[0] > 1 and wide_corr.shape[1] > 1:
        # 5. Матрица корреляций Пирсона
        corr_matrix = wide_corr.corr()
        
        # 6. Короткие подписи для тепловой карты (как было у пользователя)
        labels_map = {
            'id_1.1': 'ПЗ всего',
            'id_2.1.4': 'ОЗ всего',
            'id_3.8': 'Смертность',
            'id_5.4': 'Излечение',
            'id_5.5': 'Абацилирование',
            'id_4.2': 'МЛУ (частота)',
            'id_6.1': 'Госпитализация ВВ',
            'id_7.3': 'Охват ФЛГ',
            'id_2.4.4': 'Доля ФКТ',
            'id_2.5.4': 'Доля распада',
            'id_2.9.1': 'Охват ВИЧ (ВВ)',
            'id_2.10.1': 'ПЗ ТБ+ВИЧ'
        }
        
        # 7. Тепловая карта (короткие подписи)
        plt.figure(figsize=(12, 10))
        mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
        sns.heatmap(corr_matrix, mask=mask, annot=True, fmt='.2f', cmap='coolwarm',
                    square=True, linewidths=0.5, cbar_kws={"shrink": 0.8})
        plt.title('Корреляционная матрица ключевых показателей туберкулёза (по годам)', fontsize=14)
        
        new_labels = [labels_map.get(id_, id_) for id_ in wide_corr.columns]
        plt.gca().set_xticklabels(new_labels, rotation=45, ha='right')
        plt.gca().set_yticklabels(new_labels, rotation=0)
        plt.tight_layout()
        plt.show()
        
        # 8. Словарь id -> полное название из поля name (для вывода в консоль)
        id_to_fullname = df_indicators_sub[['id', 'name']].drop_duplicates('id').set_index('id')['name'].to_dict()
        
        # 9. Вывод сильных корреляций с ПОЛНЫМИ названиями
        print("\n=== Сильные корреляции (|r| > 0.7) ===")
        strong_corr = []
        for i in range(len(corr_matrix.columns)):
            for j in range(i+1, len(corr_matrix.columns)):
                r = corr_matrix.iloc[i, j]
                if abs(r) > 0.7:
                    strong_corr.append((corr_matrix.columns[i], corr_matrix.columns[j], r))
        if strong_corr:
            for a, b, r in strong_corr:
                name_a = id_to_fullname.get(a, a)
                name_b = id_to_fullname.get(b, b)
                print(f"{name_a} ↔ {name_b}: {r:.2f}")
        else:
            print("Нет корреляций выше 0.7 по модулю")
    else:
        print("Недостаточно данных для построения корреляционной матрицы")
else:
    print("Недостаточно ID для корреляционного анализа")
```

### Выбросы (Z-score) по возрастным группам

```python
# 8. Выбросы (на примере первичной заболеваемости по возрастам)
age_indicators = df_indicators_sub[(df_indicators_sub['type'] == 'Заболеваемость') &
                                   (df_indicators_sub['subtype'] == 'первичная') &
                                   (df_indicators_sub['population'].isin(age_groups))].copy()

if not age_indicators.empty:
    age_indicators = age_indicators.drop_duplicates(subset=['year', 'population'])
    age_indicators['zscore'] = age_indicators.groupby('population')['indicator'].transform(
        lambda x: (x - x.mean()) / x.std() if x.std() != 0 else 0
    )
    outliers = age_indicators[age_indicators['zscore'].abs() > 2]
    print(f"\nКоличество выбросов (|z|>2) по возрастам: {len(outliers)}")
    if len(outliers) > 0:
        print(outliers[['year', 'population', 'indicator', 'zscore']].head(10))
```

`Для возрастной группы 7-14 лет:`

Значение заболеваемости 17.43 в 2016 году значительно выше среднего для этой возрастной группы по всем годам

Z-score = 2.49 означает, что это значение находится на 2.49 стандартных отклонения выше среднего

Это действительная аномалия (выходит за пределы 95% доверительного интервала)

`Для возрастной группы 15-17 лет:`

Значение 44.01 в 2016 году также выше среднего

Z-score = 2.19 — тоже выброс, хотя менее сильный

Возможные причины таких выбросов:
Реальная эпидемиологическая вспышка в 2016 году среди подростков и детей

Изменение методов диагностики или учёта в 2016 году (например, более активное выявление)

Особенности формирования выборки (например, в 2016 году могли быть включены дополнительные территории)

Ошибка в данных — возможно, в исходной таблице была опечатка или неверное значение

```python
# Подготовка данных из df_indicators_sub
data_list = []

for pop in ['7-14', '15-17']:
    df_pop = df_indicators_sub[
        (df_indicators_sub['population'] == pop) &
        (df_indicators_sub['type'] == 'Заболеваемость') &
        (df_indicators_sub['subtype'] == 'первичная') &
        (df_indicators_sub['disease'] == 'по форме 8')
    ].copy()
    df_pop = df_pop[['year', 'indicator']].sort_values('year')
    df_pop['group'] = pop
    data_list.append(df_pop)

# Объединяем данные
df_plot = pd.concat(data_list)

# Функция для определения аномалий (Z-score > 2 или < -2)
def detect_outliers(data):
    if len(data) < 3:
        return []
    mean = data.mean()
    std = data.std()
    if std == 0:
        return []
    z_scores = (data - mean) / std
    return data[abs(z_scores) > 2]

# Построение графика
plt.figure(figsize=(12, 6))

for group in ['7-14', '15-17']:
    data = df_plot[df_plot['group'] == group].sort_values('year')
    years = data['year'].values
    values = data['indicator'].values
    
    # Основная линия
    plt.plot(years, values, marker='o', label=group, linewidth=2)
    
    # Поиск и выделение аномалий
    outliers = detect_outliers(pd.Series(values))
    if len(outliers) > 0:
        outlier_years = years[np.isin(values, outliers.values)]
        outlier_values = values[np.isin(values, outliers.values)]
        plt.scatter(outlier_years, outlier_values, color='red', s=120, zorder=5, edgecolors='darkred', linewidth=1.5)
        # Подписи аномалий
        for y, val in zip(outlier_years, outlier_values):
            plt.annotate(f'{val:.1f}', (y, val), textcoords="offset points", xytext=(20, 0), ha='center', fontsize=8, color='red')

plt.xlabel('Год')
plt.ylabel('Первичная заболеваемость (на 100 тыс.)')
plt.title('Динамика первичной заболеваемости туберкулёзом\n(красные точки — статистические аномалии)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.xticks(sorted(df_plot['year'].unique()))
plt.tight_layout()
plt.show()
```

```python
# 9. Распределение значений по типу показателя (boxplot) - горизонтальный
plt.figure(figsize=(12, 8))
data_for_box = df_indicators_sub[df_indicators_sub['indicator'].notna()].copy()
top_types = data_for_box['type'].value_counts().head(10).index
data_for_box = data_for_box[data_for_box['type'].isin(top_types)]

# Создаём boxplot
ax = sns.boxplot(data=data_for_box, y='type', x='indicator')

# Настройка внешнего вида
ax.set_xlabel('Значение показателя', fontsize=12)
ax.set_ylabel('Тип показателя', fontsize=12)
ax.set_title('Распределение значений indicator по типам показателей', fontsize=14, fontweight='bold')

# Устанавливаем сетку с шагом 100
ax.set_xticks(range(0, int(ax.get_xlim()[1]) + 100, 100))
ax.grid(True, linestyle='--', alpha=0.6, axis='x')

# Жирные подписи на осях
ax.tick_params(axis='both', labelsize=10)

plt.tight_layout()
plt.show()
```
