# Часть 4. Аналитический классификатор эпидпоказателей туберкулёза (АКЭПТ)

Детальная классификация абсолютно оправдано с точки зрения последующего построения дашбордов. Вот ключевые аргументы для коллег-аналитиков:

1. **Единое «окно» в разнородные данные**  
   Исходные таблицы (1000, 2100, 2400 и т.д.) имеют разную структуру, названия колонок и уровни агрегации. Классификация приводит их к единому формату, что позволяет без дополнительных преобразований объединять, фильтровать и сравнивать показатели из разных форм внутри одного дашборда.

2. **Гибкая фильтрация и агрегация**  
   Благодаря отдельным полям «type», «subtype», «disease», «people» аналитик может:  
   - оставить только нужный тип (например, «Заболевания» или «Смертность»);  
   - детализировать подтип («впервые», «рецидив», «на учёте»);  
   - сфокусироваться на конкретной форме болезни («легочный», «внелегочный», «туберкулез+ВИЧ»);  
   - выделить определённую популяцию (дети, взрослые, сельские жители).  
   Это позволяет строить дашборды с любым уровнем «drill‑down» – от общих показателей до узких разрезов.

3. **Сокращение ошибок и неоднозначности**  
   Без классификации один и тот же смысловой показатель мог бы выглядеть в разных таблицах по-разному. Чёткие правила отнесения к категориям исключают разночтения и гарантируют, что, например, «первичная заболеваемость» везде будет обозначена одинаково.

4. **Удобство для когортного и динамического анализа**  
   Наличие поля «year» вместе с категориями позволяет отслеживать изменения во времени внутри каждой фиксированной группы. Это критично для построения трендов, сравнения сезонов и оценки эффективности мер.

5. **Масштабируемость**  
   При добавлении новых таблиц или показателей классификация легко расширяется – достаточно определить для нового показателя четыре метки по тем же правилам, и он автоматически впишется в существующую аналитическую модель.

Таким образом, дробление на подкатегории – это не самоцель, а **инструмент, превращающий «сырые» статистические таблицы в готовую к анализу структурированную базу**. Затраты на первоначальную классификацию многократно окупаются скоростью и точностью разработки дашбордов.

#### Уникальные нетехнических столбцов


```python
# ======================================================================
# ВЫВОД УНИКАЛЬНЫХ ЗНАЧЕНИЙ ДЛЯ ВСЕХ НЕТЕХНИЧЕСКИХ КОЛОНОК
# (исключаем: Год, Значение, Строка, Графа, Код по МКБ, id, year, value, row, column, mkb_code, table)
# ======================================================================

df_all_names = [
    'df_all_1000', 'df_all_1100', 'df_all_2100', 'df_all_2200', 'df_all_2300',
    'df_all_2310', 'df_all_2400', 'df_all_2500', 'df_all_2513', 'df_all_2600',
    'df_all_2700', 'df_all_2800', 'df_all_3100'
]

# Колонки, которые нужно исключить из вывода (технические)
exclude_cols = [
    'Год', 'Значение', 'Строка', 'Графа', 'Код по МКБ-X пересмотра', 'Код по МКБ-Х пересмотра',
    'id', 'year', 'value', 'row', 'column', 'mkb_code', 'table', 'Таблица'
    
]

for name in df_all_names:
    df = globals()[name]
    print(f"\n{'='*80}")
    print(f"📊 {name}")
    print(f"{'='*80}")
    
    # Отбираем колонки, которые не входят в исключения
    meaningful_cols = [col for col in df.columns if col not in exclude_cols and not col.startswith('_')]
    
    for col in meaningful_cols:
        if df[col].dtype == 'object' or df[col].dtype.name == 'category':
            unique_vals = df[col].dropna().unique()
            print(f"\n🔹 {col} ({len(unique_vals)} уникальных значений):")
            if len(unique_vals) <= 20:
                for v in sorted(unique_vals):
                    print(f"   • {v}")
            else:
                for v in sorted(unique_vals)[:20]:
                    print(f"   • {v}")
                print(f"   ... и еще {len(unique_vals)-20} значений")
```

```python
# ======================================================================
# ВСЕ СТОЛБЦЫ ДЛЯ КОНКРЕТНОЙ ГРУППЫ ДУБЛИКАТОВ
# ======================================================================

df = df_all_1000.copy()

# Берем первую группу дубликатов из вашего вывода
# Формы туберкулеза = 'Из общего числа больных (стр. 01,02)', Пол = 'М', Возраст = 'всего', Год = 2016
mask = (
    (df['Формы туберкулеза'] == 'Из общего числа больных (стр. 01,02)') &
    (df['Пол'] == 'М') &
    (df['Возраст'] == 'всего') &
    (df['Год'] == 2016)
)

duplicates = df[mask]

print(f"Найдено {len(duplicates)} строк-дубликатов")
print("\n" + "="*80)
print("ВСЕ СТОЛБЦЫ ДЛЯ ЭТИХ ДУБЛИКАТОВ:")
print("="*80)
display(duplicates)
```

```python
# ======================================================================
# УДАЛЕНИЕ ДУБЛИКАТОВ ПО (Строка, Графа, Год) В df_all_1000
# ======================================================================

df = df_all_1000.copy()

before = len(df)

# Удаляем дубликаты, оставляя первое вхождение
df = df.drop_duplicates(subset=['Строка', 'Графа', 'Год'], keep='first')

after = len(df)

print(f"✅ Было строк: {before}")
print(f"✅ Стало строк: {after}")
print(f"✅ Удалено дубликатов: {before - after}")

# Сохраняем обратно
df_all_1000 = df

print("\nПроверка:")
dup_check = df.duplicated(subset=['Строка', 'Графа', 'Год']).sum()
print(f"Осталось дубликатов: {dup_check}")
```

```python
# ======================================================================
# СОЗДАНИЕ КОЛОНКИ 'name' ПУТЁМ ОБЪЕДИНЕНИЯ ОПРЕДЕЛЯЮЩИХ КОЛОНОК
# ======================================================================

# df_all_1000: Формы туберкулеза + Пол + Возраст
df_all_1000['name'] = (
    df_all_1000['Формы туберкулеза'].fillna('').astype(str) + ' | ' +
    df_all_1000['Пол'].fillna('').astype(str) + ' | ' +
    df_all_1000['Возраст'].fillna('').astype(str)
)

# df_all_1100: Должность
df_all_1100['name'] = df_all_1100['Должность'].fillna('').astype(str)

# df_all_2100: Формы туберкулеза + Уточнение + Еще уточнение
df_all_2100['name'] = (
    df_all_2100['Формы туберкулеза'].fillna('').astype(str) + ' | ' +
    df_all_2100['Уточнение'].fillna('').astype(str) + ' | ' +
    df_all_2100['Еще уточнение'].fillna('').astype(str)
)

# df_all_2200: Наименование показателя + Уточнение
df_all_2200['name'] = (
    df_all_2200['Наименование показателя'].fillna('').astype(str) + ' | ' +
    df_all_2200['Уточнение'].fillna('').astype(str)
)

# df_all_2300: Показатель + Уточнение + Еще уточнение
df_all_2300['name'] = (
    df_all_2300['Показатель'].fillna('').astype(str) + ' | ' +
    df_all_2300['Уточнение'].fillna('').astype(str) + ' | ' +
    df_all_2300['Еще уточнение'].fillna('').astype(str)
)

# df_all_2310: Показатель + Уточнение + Еще уточнение
df_all_2310['name'] = (
    df_all_2310['Показатель'].fillna('').astype(str) + ' | ' +
    df_all_2310['Уточнение'].fillna('').astype(str) + ' | ' +
    df_all_2310['Еще уточнение'].fillna('').astype(str)
)

# df_all_2400: Показатель + Уточнение
df_all_2400['name'] = (
    df_all_2400['Показатель'].fillna('').astype(str) + ' | ' +
    df_all_2400['Уточнение'].fillna('').astype(str)
)

# df_all_2500: Показатель + Уточнение
df_all_2500['name'] = (
    df_all_2500['Показатель'].fillna('').astype(str) + ' | ' +
    df_all_2500['Уточнение'].fillna('').astype(str)
)

# df_all_2513: Профилактические осмотры на туберкулез + Категория
df_all_2513['name'] = (
    df_all_2513['Профилактические осмотры на туберкулез'].fillna('').astype(str) + ' | ' +
    df_all_2513['Категория'].fillna('').astype(str)
)

# df_all_2600: Показатель + Возраст
df_all_2600['name'] = (
    df_all_2600['Показатель'].fillna('').astype(str) + ' | ' +
    df_all_2600['Возраст'].fillna('').astype(str)
)

# df_all_2700: Показатель + Уточнение
df_all_2700['name'] = (
    df_all_2700['Показатель'].fillna('').astype(str) + ' | ' +
    df_all_2700['Уточнение'].fillna('').astype(str)
)

# df_all_2800: Показатель + Возраст
df_all_2800['name'] = (
    df_all_2800['Показатель'].fillna('').astype(str) + ' | ' +
    df_all_2800['Возраст'].fillna('').astype(str)
)

# df_all_3100: Тип_коек
df_all_3100['name'] = df_all_3100['Тип_коек'].fillna('').astype(str)

print("✅ Колонка 'name' добавлена во все датасеты. Старые колонки сохранены.")

```


## Классификация показателей во всех таблицах

Процесс

Для каждого датасета `df_all_*` добавлены единые колонки:

- **`type`** – тип показателя (Заболевания, Умерло, Госпитализация, Кадры и др.)
- **`subtype`** – детализация (впервые, рецидив, общая, III группа и т.п.)
- **`disease`** – форма/локализация туберкулёза (лёгочный, внелегочный, ФКТ, ВИЧ и др.)
- **`people`** – популяционная группа (возраст, пол, социальный статус)

Принципы классификации

1. **Таблицы с уточнениями** (2100, 2200, 2300, 2400, 2500, 2700):  
   `type` и `disease` – из основного показателя; `subtype` и `people` – из уточнений (Возраст, Категория и др.).

2. **Таблицы без уточнений** (1000, 1100, 2600, 2800, 3100):  
   Все категории – по основному показателю. Для 1000 `people` формируется из пола и возраста.

3. **Таблицы с множественными уточнениями** (2310, 2513):  
   Все уточнения объединяются в единый ключ, классификация – по полному словарю комбинаций.

Примеры

| Таблица | Пример | type | subtype | disease | people |
|---------|--------|------|---------|---------|------------|
| 1000 | *Впервые выявленный активный туберкулёз, всего* | Заболевания | впервые | активный туберкулёз | всего |
| 2100 | *Туберкулёз органов дыхания + Взято на учёт* | Заболевания | впервые | туберкулёз органов дыхания | 0-14 |
| 2400 | *Взрослые, группа 0А + Снято* | Диспансерный учёт (0А) | снято | туберкулёз (акт. не опр.) | 18+ |

Итоговые типы показателей

| type | Примеры |
|------|---------|
| Заболевания | впервые выявленные, рецидивы |
| Умерло | от туберкулёза, от других причин |
| Госпитализация | всего, БК+, дневной стационар |
| Хирургическое лечение | органы дыхания, ФКТ, костно-суставной |
| Диспансерный учёт | 0А, III, IVA, VIA, VIБ |
| Движение контингентов | прибыло, выбыло, переведено |
| Лекарственная устойчивость | впервые выявленные, I группа учёта |
| Профилактические осмотры | флюорография, иммунодиагностика |
| Кадры, Коечный фонд, Выявление, Наблюдение, Лечение, Инвалидность, Обследовано, Лабораторная диагностика | – |

### Унификация названий столбцов

```python
# ======================================================================
# ПЕРЕИМЕНОВАНИЕ ВСЕХ ОСТАВШИХСЯ РУССКИХ НАЗВАНИЙ КОЛОНОК В АНГЛИЙСКИЕ
# ======================================================================

# Единый словарь для переименования русских колонок в английские (во всех датасетах)
final_rename = {
    'Формы туберкулеза': 'tb_form',
    'Должность': 'position',
    'Наименование показателя': 'row_name',
    'Показатель': 'row_name',
    'Профилактические осмотры на туберкулез': 'preventive_exam',
    'Категория': 'category',
    'Код по МКБ-X пересмотра': 'mkb_code',
    'Код по МКБ-Х пересмотра': 'mkb_code',
    'Уточнение': 'clarification',
    'Еще уточнение': 'clarification2',
    'Возраст': 'age',
    'Пол': 'gender',
    'Тип_коек': 'bed_type',
    'Строка': 'row',
    'Графа': 'column',
    'Год': 'year',
    'Таблица': 'table',
    'Значение': 'value',
}

# Список датасетов
df_all_names = [
    'df_all_1000', 'df_all_1100', 'df_all_2100', 'df_all_2200', 'df_all_2300',
    'df_all_2310', 'df_all_2400', 'df_all_2500', 'df_all_2513', 'df_all_2600',
    'df_all_2700', 'df_all_2800', 'df_all_3100'
]

# Применяем переименование
for name in df_all_names:
    df = globals()[name]
    # Переименовываем только те колонки, которые есть в df и есть в словаре
    existing_rename = {k: v for k, v in final_rename.items() if k in df.columns}
    if existing_rename:
        df = df.rename(columns=existing_rename)
        globals()[name] = df
        print(f"✓ Переименованы колонки в {name}")
    else:
        print(f"⚠️ В {name} нет русских колонок для переименования")

print("\n" + "="*80)
print("КОЛОНКИ ПОСЛЕ ФИНАЛЬНОГО ПЕРЕИМЕНОВАНИЯ")
print("="*80)

for name in df_all_names:
    df = globals()[name]
    print(f"\n{name}: {list(df.columns)}")
```

### Классификация маппингом

```python
# ======================================================================
# МАППИНГ КЛАССИФИКАЦИИ ДЛЯ ВСЕХ ТАБЛИЦ С СОЗДАНИЕМ df_map_{table}
# (СОХРАНЯЕМ ВСЕ КОЛОНКИ)
# ======================================================================

# Список датасетов и соответствующих словарей
mapping_config = [
    ('df_all_1000', class_dict_1000, 'df_map_1000'),
    #('df_all_1100', class_dict_1100, 'df_map_1100'),
    ('df_all_2100', class_dict_2100, 'df_map_2100'),
    ('df_all_2200', class_dict_2200, 'df_map_2200'),
    ('df_all_2300', class_dict_2300, 'df_map_2300'),
    ('df_all_2310', class_dict_2310, 'df_map_2310'),
    ('df_all_2400', class_dict_2400, 'df_map_2400'),
    ('df_all_2500', class_dict_2500, 'df_map_2500'),
    ('df_all_2513', class_dict_2513, 'df_map_2513'),
    ('df_all_2600', class_dict_2600, 'df_map_2600'),
    ('df_all_2700', class_dict_2700, 'df_map_2700'),
    ('df_all_2800', class_dict_2800, 'df_map_2800')
    #('df_all_3100', class_dict_3100, 'df_map_3100'),
]

print("="*80)
print("СОЗДАНИЕ df_map_* ТАБЛИЦ (СО ВСЕМИ КОЛОНКАМИ)")
print("="*80)

for source_name, class_dict, target_name in mapping_config:
    df_source = globals()[source_name]
    source_rows = len(df_source)
    
    # СОХРАНЯЕМ ВСЕ КОЛОНКИ
    df_map = df_source.copy()
    
    # Добавляем колонки классификации
    df_map['type'] = df_map['name'].map(lambda x: class_dict.get(x, ('', '', '', ''))[0])
    df_map['subtype'] = df_map['name'].map(lambda x: class_dict.get(x, ('', '', '', ''))[1])
    df_map['disease'] = df_map['name'].map(lambda x: class_dict.get(x, ('', '', '', ''))[2])
    df_map['people'] = df_map['name'].map(lambda x: class_dict.get(x, ('', '', '', ''))[3])
    
    # Удаляем строки с пустыми значениями (если есть)
    df_map = df_map.dropna(subset=['type', 'subtype', 'disease', 'people'])
    
    globals()[target_name] = df_map
    
    print(f"\n📊 {source_name}: {source_rows} строк")
    print(f"   → {target_name}: {len(df_map)} строк, колонки: {list(df_map.columns)}")

print("\n" + "="*80)
print("ВСЕ df_map_* СОЗДАНЫ")
print("="*80)

# Проверка
print("\n=== df_map_1000 (первые 5 строк, все колонки) ===")
display(df_map_1000.head(5))
```

## Формирование `id`
 Алгоритм формирования

1. **Номер таблицы** – извлекается из имени датасета (например, `df_all_1000` → `1000`, `df_all_2513` → `2513`)

2. **Для каждой классификационной колонки** (`type`, `subtype`, `disease`, `people`):
   - Собираются все уникальные значения
   - Значения сортируются в алфавитном порядке
   - Каждому значению присваивается порядковый номер (начиная с 1)

3. **Для каждой строки** – берутся соответствующие номера и формируется `id`

 Пример для `df_all_1000`

**Уникальные значения:**
| Колонка | Уникальные значения | Номера |
|---------|--------------------|--------|
| `type` | `['Заболевания']` | `Заболевания` → 1 |
| `subtype` | `['первичная', 'рецидив']` | `первичная` → 1, `рецидив` → 2 |
| `disease` | `['ФКТ', 'активный туберкулез', 'туберкулез', ...]` | нумерация по алфавиту |
| `people` | `['0-4', '15-17', '18-24', ..., 'всего', 'женщины 0-4', ...]` | нумерация по алфавиту |

**Результат для строки:**
- `type` = `'Заболевания'` → номер 1
- `subtype` = `'первичная'` → номер 1  
- `disease` = `'туберкулез'` → номер 2 (допустим, после `'ФКТ'` = 1, `'активный туберкулез'` = 2?)
- `people` = `'женщины 65+'` → номер 33

**Полученный `id`:**

```python
# ======================================================================
# ДОБАВЛЕНИЕ КОРОТКОГО ID ДЛЯ df_map_* ТАБЛИЦ
# ======================================================================

df_map_names = [
    'df_map_1000', 
    #'df_map_1100', 
    'df_map_2100', 'df_map_2200', 'df_map_2300',
    'df_map_2310', 'df_map_2400', 'df_map_2500', 'df_map_2513', 'df_map_2600',
    'df_map_2700', 'df_map_2800'
    #'df_map_3100'
]

# Извлекаем номер таблицы из имени датасета
table_numbers = {
    'df_map_1000': '1000',
#  'df_map_1100': '1100',
    'df_map_2100': '2100',
    'df_map_2200': '2200',
    'df_map_2300': '2300',
    'df_map_2310': '2310',
    'df_map_2400': '2400',
    'df_map_2500': '2500',
    'df_map_2513': '2513',
    'df_map_2600': '2600',
    'df_map_2700': '2700',
    'df_map_2800': '2800'
#    'df_map_3100': '3100'
}

for name in df_map_names:
    if name not in globals():
        print(f"⚠️ {name} не найден в globals(), пропускаем")
        continue
    
    df = globals()[name].copy()
    
    # Удаляем существующий id
    if 'id' in df.columns:
        df = df.drop(columns=['id'])
    
    # Создаём словари для нумерации
    type_map = {val: i+1 for i, val in enumerate(sorted(df['type'].unique()))}
    subtype_map = {val: i+1 for i, val in enumerate(sorted(df['subtype'].unique()))}
    disease_map = {val: i+1 for i, val in enumerate(sorted(df['disease'].unique()))}
    people_map = {val: i+1 for i, val in enumerate(sorted(df['people'].unique()))}
    
    # Формируем id
    def make_id(row):
        return f"id_{table_numbers[name]}_{type_map[row['type']]}_{subtype_map[row['subtype']]}_{disease_map[row['disease']]}_{people_map[row['people']]}"
    
    df.insert(0, 'id', df.apply(make_id, axis=1))
    
    globals()[name] = df
    print(f"✓ {name}: {len(df)} строк, уникальных type={df['type'].nunique()}, subtype={df['subtype'].nunique()}, disease={df['disease'].nunique()}, people={df['people'].nunique()}")

print("\n✅ Все id добавлены!")

# Проверка для df_map_2700
print("\n=== df_map_2700 (первые 10 строк) ===")
if 'df_map_2700' in globals():
    display(df_map_2700[['id', 'type', 'subtype', 'disease', 'people', 'name']].head(10))
else:
    print("df_map_2700 не найден")
```

## Проверка уникальности id

```python
# ======================================================================
# ПРОВЕРКА ДУБЛИКАТОВ ПО (type, subtype, disease, people, year)
# ======================================================================

df_map_names = [
    'df_map_1000', #'df_map_1100', 
    'df_map_2100', 'df_map_2200', 'df_map_2300',
    'df_map_2310', 'df_map_2400', 'df_map_2500', 'df_map_2513', 'df_map_2600',
    'df_map_2700', 'df_map_2800'
    #'df_map_3100'
]

print("="*80)
print("ПРОВЕРКА ДУБЛИКАТОВ ПО (type, subtype, disease, people, year)")
print("="*80)

for name in df_map_names:
    if name not in globals():
        print(f"\n⚠️ {name}: не найден")
        continue
    
    df = globals()[name]
    
    # Проверяем наличие колонок
    required = ['type', 'subtype', 'disease', 'people', 'year']
    if not all(col in df.columns for col in required):
        print(f"\n⚠️ {name}: нет колонок {required}")
        continue
    
    # Считаем дубликаты
    dup_count = df.duplicated(subset=required).sum()
    
    if dup_count == 0:
        print(f"\n✅ {name}: нет дубликатов")
    else:
        print(f"\n❌ {name}: найдено {dup_count} дубликатов")
        
        # Показываем примеры
        duplicates = df[df.duplicated(subset=required, keep=False)]
        print(f"\n   Пример (первые 3):")
        for _, row in duplicates.head(3).iterrows():
            print(f"   type={row['type']}, subtype={row['subtype']}, disease={row['disease']}, people={row['people']}, year={row['year']}")
            print(f"   name: {row['name'][:60]}...")
            print()
```

## Порядок столбцов

```python
# ======================================================================
# ИЗМЕНЕНИЕ ПОРЯДКА СТОЛБЦОВ ДЛЯ ВСЕХ df_map_* ДАТАСЕТОВ (С ID В НАЧАЛЕ)
# ======================================================================

# Базовый порядок колонок
base_order = ['id', 'name', 'type', 'subtype', 'disease', 'people', 'year', 'value']

# Дополнительные колонки для каждого датасета
extra_columns = {
    'df_map_1000': ['tb_form', 'gender', 'age', 'mkb_code', 'row', 'column', 'table'],
    'df_map_2100': ['tb_form', 'clarification', 'clarification2', 'mkb_code', 'row', 'column', 'table'],
    'df_map_2200': ['row_name', 'clarification', 'row', 'column', 'table'],
    'df_map_2300': ['row_name', 'clarification', 'clarification2', 'row', 'column', 'table'],
    'df_map_2310': ['row_name', 'clarification', 'clarification2', 'row', 'column', 'table'],
    'df_map_2400': ['row_name', 'clarification', 'row', 'column', 'table'],
    'df_map_2500': ['row_name', 'clarification', 'row', 'column', 'table'],
    'df_map_2513': ['preventive_exam', 'category', 'row', 'column', 'table'],
    'df_map_2600': ['row_name', 'age', 'row', 'column', 'table'],
    'df_map_2700': ['row_name', 'clarification', 'row', 'column', 'table'],
    'df_map_2800': ['row_name', 'age', 'row', 'column', 'table'],
}

df_map_names = [
    'df_map_1000', 'df_map_2100', 'df_map_2200', 'df_map_2300',
    'df_map_2310', 'df_map_2400', 'df_map_2500', 'df_map_2513', 
    'df_map_2600', 'df_map_2700', 'df_map_2800'
]

for name in df_map_names:
    if name not in globals():
        print(f"⚠️ {name}: не найден")
        continue
    
    df = globals()[name]
    
    # Формируем порядок колонок
    new_order = []
    
    # Добавляем базовые колонки
    for col in base_order:
        if col in df.columns:
            new_order.append(col)
    
    # Добавляем дополнительные колонки
    if name in extra_columns:
        for col in extra_columns[name]:
            if col in df.columns:
                new_order.append(col)
    
    # Добавляем остальные колонки
    for col in df.columns:
        if col not in new_order:
            new_order.append(col)
    
    # Переставляем колонки
    df = df[new_order]
    globals()[name] = df
    
    print(f"✅ {name}: порядок колонок изменён ({len(new_order)} колонок)")

print("\n✅ Порядок колонок обновлён для всех датасетов!")

# Проверка для df_map_1000
print("\n📋 df_map_1000 (первые 10 колонок):")
print(list(df_map_1000.columns[:10]))
```

```python
# ======================================================================
# ВЫВОД НАЗВАНИЙ ВСЕХ КОЛОНОК ДЛЯ КАЖДОГО df_map_* ПОСЛЕ ПЕРЕСТАНОВКИ
# ======================================================================

df_map_names = [
    'df_map_1000', 'df_map_2100', 'df_map_2200', 'df_map_2300',
    'df_map_2310', 'df_map_2400', 'df_map_2500', 'df_map_2513', 
    'df_map_2600', 'df_map_2700', 'df_map_2800'
]

print("="*80)
print("СПИСОК КОЛОНОК ПОСЛЕ ПЕРЕСТАНОВКИ ПОРЯДКА")
print("="*80)

for name in df_map_names:
    if name not in globals():
        print(f"\n⚠️ {name}: не найден")
        continue
    
    df = globals()[name]
    columns = list(df.columns)
    
    print(f"\n📊 {name}")
    print(f"   Всего колонок: {len(columns)}")
    print(f"   Колонки: {columns}")
```

```python
# ======================================================================
# ПЕРЕИМЕНОВАНИЕ ТАБЛИЦ df_map_* В ЧИТАЕМЫЕ НАЗВАНИЯ
# ======================================================================

rename_dict = {
    'df_map_1000': '1000 Число больных с впервые в жизни установленным диагнозом активного туберкулеза',
    'df_all_1100': '1100 Должности и физические лица медицинской организации',
    'df_map_2100': '2100 Контингенты больных активным туберкулезом, состоящих под наблюдением данного ЛПУ',
    'df_map_2200': '2200 Выявление больных и некоторых групп риска',
    'df_map_2300': '2300 Движение контингентов больных туберкулезом',
    'df_map_2310': '2310 Из числа умерших от туберкулеза состояло на учете менее 1 года',
    'df_map_2400': '2400 Диспансерная работа с 0, IV, VI группами учета',
    'df_map_2500': '2500 Бактериовыделители, состоящие на учете',
    'df_map_2513': '2513 Профилактические осмотры на туберкулез',
    'df_map_2600': '2600 Больничная и санаторная помощь',
    'df_map_2700': '2700 Эффективность лечения и наблюдения больных туберкулезом органов дыхания, выявленных в предыдущем году',
    'df_map_2800': '2800 Оказание помощи больными туберкулезом, временно проживающими на территории обслуживания',
    'df_all_3100': '3100 Коечный фонд и его использование',
}

# Переименовываем переменные
for old_name, new_name in rename_dict.items():
    if old_name in globals():
        globals()[new_name] = globals()[old_name]
        print(f"✓ Переименовано: {old_name} → {new_name}")
    else:
        print(f"⚠️ {old_name} не найден")

print("\n✅ Переименование завершено!")

# Проверка новых имен
print("\n📋 Доступные таблицы:")
for name in rename_dict.values():
    if name in globals():
        df = globals()[name]
        print(f"   - {name}: {len(df)} строк, {len(df.columns)} колонок")
```

## Уникальные подкатегорий

```python
# ======================================================================
# ВЫВОД ВСЕХ УНИКАЛЬНЫХ ЗНАЧЕНИЙ (type, subtype, disease, people) ДЛЯ КАЖДОГО df_map_*
# С ИСПОЛЬЗОВАНИЕМ ДЛИННЫХ НАЗВАНИЙ ТАБЛИЦ
# ======================================================================

# Словарь соответствия коротких и длинных названий
table_names = {
    'df_map_1000': '1000 Число больных с впервые в жизни установленным диагнозом активного туберкулеза',
    'df_map_1100': '1100 Должности и физические лица медицинской организации',
    'df_map_2100': '2100 Контингенты больных активным туберкулезом, состоящих под наблюдением данного ЛПУ',
    'df_map_2200': '2200 Выявление больных и некоторых групп риска',
    'df_map_2300': '2300 Движение контингентов больных туберкулезом',
    'df_map_2310': '2310 Из числа умерших от туберкулеза состояло на учете менее 1 года',
    'df_map_2400': '2400 Диспансерная работа с 0, IV, VI группами учета',
    'df_map_2500': '2500 Бактериовыделители, состоящие на учете',
    'df_map_2513': '2513 Профилактические осмотры на туберкулез',
    'df_map_2600': '2600 Больничная и санаторная помощь',
    'df_map_2700': '2700 Эффективность лечения и наблюдения больных туберкулезом органов дыхания, выявленных в предыдущем году',
    'df_map_2800': '2800 Оказание помощи больными туберкулезом, временно проживающими на территории обслуживания',
    'df_map_3100': '3100 Коечный фонд и его использование',
}

for old_name, display_name in table_names.items():
    if old_name not in globals():
        print(f"\n⚠️ {display_name}: не найден")
        continue
    
    df = globals()[old_name]
    required = ['type', 'subtype', 'disease', 'people']
    missing = [c for c in required if c not in df.columns]
    if missing:
        print(f"\n⚠️ {display_name}: нет колонок {missing}")
        continue
    
    print(f"\n{'='*100}")
    print(f"📊 {display_name}")
    print(f"{'='*100}")
    
    for col in required:
        unique_vals = sorted(df[col].dropna().unique())
        print(f"\n🔹 {col} ({len(unique_vals)} уникальных):")
        # Для длинных значений печатаем с переносом
        for i, val in enumerate(unique_vals, 1):
            if len(str(val)) > 80:
                print(f"   {i:3d}. {str(val)[:77]}...")
            else:
                print(f"   {i:3d}. {val}")
    
    print(f"\n{'-'*100}")
    print(f"📋 Примеры строк (первые 3):")
    display(df[['name', 'type', 'subtype', 'disease', 'people']].head(3))
```

## Сохранение

```python
df_map_1100 = df_all_1100
df_map_3100 = df_all_3100

# Список датасетов df_map_*
df_map_names = [
    'df_map_1000', 'df_map_2100', 'df_map_2200', 'df_map_2300',
    'df_map_2310', 'df_map_2400', 'df_map_2500', 'df_map_2513', 
    'df_map_2600', 'df_map_2700', 'df_map_2800', 'df_map_1100', 'df_map_3100'
]

# Функция для создания ссылки на скачивание
def create_download_link(df, filename):
    csv = df.to_csv(index=False, encoding='utf-8-sig')
    b64 = base64.b64encode(csv.encode()).decode()
    display(HTML(f'<a href="data:file/csv;base64,{b64}" download="{filename}">Скачать {filename}</a>'))

# Сохраняем каждый df_map_* в CSV и выводим ссылки
print("Сохранение df_map_* датасетов в CSV и ссылки для скачивания:")
for name in df_map_names:
    if name in globals():
        df = globals()[name]
        filename = f'{name}.csv'
        df.to_csv(filename, index=False, encoding='utf-8-sig')  # сохраняем на диск
        create_download_link(df, filename)                      # ссылка для скачивания
        print(f"✓ {filename} сохранён и доступен по ссылке выше")
    else:
        print(f"⚠️ {name} не найден в globals()")

print("\n✅ Готово! Все файлы сохранены в текущую директорию и доступны для скачивания.")
```

```python
# Функция для создания ссылки на скачивание
def create_download_link(df, filename):
    csv = df.to_csv(index=False, encoding='utf-8-sig')
    b64 = base64.b64encode(csv.encode()).decode()
    display(HTML(f'<a href="data:file/csv;base64,{b64}" download="{filename}">📥 Скачать {filename}</a>'))

# Список переименованных таблиц
renamed_tables = [
    '1000 Число больных с впервые в жизни установленным диагнозом активного туберкулеза',
    '1100 Должности и физические лица медицинской организации',
    '2100 Контингенты больных активным туберкулезом, состоящих под наблюдением данного ЛПУ',
    '2200 Выявление больных и некоторых групп риска',
    '2300 Движение контингентов больных туберкулезом',
    '2310 Из числа умерших от туберкулеза состояло на учете менее 1 года',
    '2400 Диспансерная работа с 0, IV, VI группами учета',
    '2500 Бактериовыделители, состоящие на учете',
    '2513 Профилактические осмотры на туберкулез',
    '2600 Больничная и санаторная помощь',
    '2700 Эффективность лечения и наблюдения больных туберкулезом органов дыхания, выявленных в предыдущем году',
    '2800 Оказание помощи больными туберкулезом, временно проживающими на территории обслуживания',
    '3100 Коечный фонд и его использование',
]

print("="*80)
print("ССЫЛКИ ДЛЯ СКАЧИВАНИЯ ТАБЛИЦ")
print("="*80)
print()

for table_name in renamed_tables:
    if table_name in globals():
        df = globals()[table_name]
        filename = f"{table_name}.csv"
        create_download_link(df, filename)
    else:
        print(f"   ❌ {table_name} не найден\n")
```

## Вывод

Создана унифицированная классификационная модель, которая приводит 11 таблиц абсолютных показателей (формы 1000, 2100, 2200, 2300, 2310, 2400, 2500, 2513, 2600, 2700, 2800) к единой аналитической структуре с колонками `type`, `subtype`, `disease`, `people`, `year` и `value`.  
Все показатели классифицированы по типам (Заболевания, Умерло, Госпитализация, Диспансерный учет и т.д.), подтипам (впервые, рецидив, на учёте, впервые выявлен ТБ и др.), форме заболевания и популяции (возраст/пол/социальная группа).  
Такая структура позволяет:

- **Объединять и сравнивать данные** из разных источников без дополнительных преобразований.
- **Гибко фильтровать и агрегировать** показатели для построения дашбордов (от общего тренда до детализированных срезов).
- **Исключить неоднозначность** интерпретации показателей за счёт жёсткой классификации.
- **Строить когортные и динамические отчёты** с привязкой к году.

В результате получен готовый к использованию аналитический слой, на котором можно быстро разрабатывать визуализации и проводить эпидемиологический анализ абсолютных показателей туберкулёза.
